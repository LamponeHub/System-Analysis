```python
# Ячейка 0: Импорты
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
# Задания 1-3
calls = pd.read_csv('/content/datasets/calls.csv')
display(calls.head())
calls.info()
# Задание 4
calls['duration'].hist(bins=30, figsize=(10, 5), edgecolor='black')
plt.title('Распределение продолжительности звонков')
plt.xlabel('Минуты')
plt.ylabel('Количество звонков')
plt.show()
# Задания 5-7
sessions = pd.read_csv('/content/datasets/internet.csv')
display(sessions.head())
sessions.info()
# Задание 8
sessions['mb_used'].hist(bins=30, figsize=(10, 5), edgecolor='black')
plt.title('Распределение использованного интернет-трафика (МБ)')
plt.xlabel('Мегабайты')
plt.ylabel('Количество сессий')
plt.show()
# Задания 9-11
messages = pd.read_csv('/content/datasets/messages.csv')
display(messages.head())
messages.info()
# Задания 12-14
tariffs = pd.read_csv('/content/datasets/tariffs.csv')
display(tariffs)
tariffs.info()
# Задания 15-17
users = pd.read_csv('/content/datasets/users.csv')
display(users.head())
users.info()
# Задание 18: Преобразование типов дат
users['reg_date'] = pd.to_datetime(users['reg_date'])
users['churn_date'] = pd.to_datetime(users['churn_date'])
calls['call_date'] = pd.to_datetime(calls['call_date'])
messages['message_date'] = pd.to_datetime(messages['message_date'])
sessions['session_date'] = pd.to_datetime(sessions['session_date'])
# Задание 19: Округление длительности звонков вверх
calls['duration'] = np.ceil(calls['duration']).astype(int)
# Задание 20: Удаление технического столбца
sessions.drop(columns='Unnamed: 0', inplace=True, errors='ignore')
# Задания 21-23: Извлечение номера месяца
calls['month'] = calls['call_date'].dt.month
messages['month'] = messages['message_date'].dt.month
sessions['month'] = sessions['session_date'].dt.month
# Задание 24: Количество звонков по пользователям и месяцам
calls_per_month = calls.groupby(['user_id', 'month']).agg({'duration': 'count'})
display(calls_per_month.head(30))
# Задание 25: Сумма минут разговора по пользователям и месяцам
minutes_per_month = calls.groupby(['user_id', 'month']).agg(minutes=('duration', 'sum'))
display(minutes_per_month.head(30))
# Задание 26: Количество сообщений по пользователям и месяцам
messages_per_month = messages.groupby(['user_id', 'month']).agg(messages=('message_date', 'count'))
display(messages_per_month.head(30))
# Задание 27: Сумма потраченных мегабайт по пользователям и месяцам
sessions_per_month = sessions.groupby(['user_id', 'month']).agg({'mb_used': 'sum'})
# Объединяем агрегированные показатели, присоединяем информацию о тарифах и рассчитываем итоговую выручку.
# Формула: абонентская плата + оплата сверх лимита (минуты, сообщения, ГБ).
# ГБ округляются вверх, как указано в условиях тарифов.
# Объединяем таблицы агрегаций
df_monthly = calls_per_month.join(minutes_per_month, how='outer')
df_monthly = df_monthly.join(messages_per_month, how='outer')
df_monthly = df_monthly.join(sessions_per_month, how='outer')
df_monthly = df_monthly.fillna(0).reset_index()

# Присоединяем данные о пользователях и тарифах
# Примечание: если в вашем датасете столбец называется 'tarif', замените 'tariff' на 'tarif'
df_monthly = df_monthly.merge(users[['user_id', 'city', 'tariff']], on='user_id', how='left')
df_monthly = df_monthly.merge(tariffs.set_index('tariff_name'), left_on='tariff', right_index=True, how='left')

# Расчёт превышения лимитов
df_monthly['extra_minutes'] = np.maximum(0, df_monthly['minutes'] - df_monthly['minutes_included'])
df_monthly['extra_messages'] = np.maximum(0, df_monthly['messages'] - df_monthly['messages_included'])

# Интернет: переводим МБ в ГБ и округляем вверх
df_monthly['used_gb'] = np.ceil(df_monthly['mb_used'] / 1024)
df_monthly['included_gb'] = df_monthly['mb_per_month_included'] / 1024
df_monthly['extra_gb'] = np.maximum(0, df_monthly['used_gb'] - df_monthly['included_gb'])

# Итоговая выручка за месяц
df_monthly['revenue'] = (df_monthly['rub_monthly_fee'] +
                         df_monthly['extra_minutes'] * df_monthly['rub_per_minute'] +
                         df_monthly['extra_messages'] * df_monthly['rub_per_message'] +
                         df_monthly['extra_gb'] * df_monthly['rub_per_gb'])

display(df_monthly[['user_id', 'month', 'tariff', 'city', 'minutes', 'messages', 'mb_used', 'extra_minutes', 'extra_messages', 'extra_gb', 'revenue']].head(10))
# Задание 28: Сравнение средней выручки тарифов «Ультра» и «Смарт»
ultra_rev = df_monthly[df_monthly['tariff'] == 'ultra']['revenue']
smart_rev = df_monthly[df_monthly['tariff'] == 'smart']['revenue']

p_val_tariff = stats.ttest_ind(ultra_rev, smart_rev, equal_var=False).pvalue
print(p_val_tariff)

if p_val_tariff < 0.05:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не получилось отвергнуть нулевую гипотезу")
    # Задание 29: Сравнение средней выручки пользователей из Москвы и других регионов
moscow_rev = df_monthly[df_monthly['city'] == 'Москва']['revenue']
other_rev = df_monthly[df_monthly['city'] != 'Москва']['revenue']

p_val_city = stats.ttest_ind(moscow_rev, other_rev, equal_var=False).pvalue
print(p_val_city)

if p_val_city < 0.05:
    print("Отвергаем нулевую гипотезу")
else:
    print("Не получилось отвергнуть нулевую гипотезу")
```
