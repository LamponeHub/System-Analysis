```python
# @title 🍷 Создание и заполнение базы данных Wine Reviews
# @markdown Этот код создаст SQLite базу данных и заполнит её примерами данных

import sqlite3
import pandas as pd
from IPython.display import display, Markdown

# Создание подключения к базе данных (в памяти)
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

print("✅ База данных создана успешно!")

# @title 📋 Создание таблиц
# @markdown Создание всех необходимых таблиц согласно схеме

# Создание таблицы countries
cursor.execute('''
CREATE TABLE countries (
    country_id INTEGER PRIMARY KEY AUTOINCREMENT,
    country_abbr TEXT,
    latitude REAL,
    longitude REAL,
    country_name TEXT,
    temperature REAL
)
''')

# Создание таблицы provinces
cursor.execute('''
CREATE TABLE provinces (
    province_id INTEGER PRIMARY KEY AUTOINCREMENT,
    province_name TEXT,
    avg_f REAL,
    avg_c REAL
)
''')

# Создание таблицы regions
cursor.execute('''
CREATE TABLE regions (
    region_id INTEGER PRIMARY KEY AUTOINCREMENT,
    region_name TEXT
)
''')

# Создание таблицы wineries
cursor.execute('''
CREATE TABLE wineries (
    winery_id INTEGER PRIMARY KEY AUTOINCREMENT,
    winery_name TEXT
)
''')

# Создание таблицы wine_types
cursor.execute('''
CREATE TABLE wine_types (
    wine_type_id INTEGER PRIMARY KEY AUTOINCREMENT,
    wine_type TEXT
)
''')

# Создание таблицы wines
cursor.execute('''
CREATE TABLE wines (
    wine_id INTEGER PRIMARY KEY AUTOINCREMENT,
    wine_name TEXT,
    country_id INTEGER,
    province_id INTEGER,
    region_id INTEGER,
    winery_id INTEGER,
    wine_type_id INTEGER,
    rating REAL,
    price REAL,
    FOREIGN KEY (country_id) REFERENCES countries(country_id),
    FOREIGN KEY (province_id) REFERENCES provinces(province_id),
    FOREIGN KEY (region_id) REFERENCES regions(region_id),
    FOREIGN KEY (winery_id) REFERENCES wineries(winery_id),
    FOREIGN KEY (wine_type_id) REFERENCES wine_types(wine_type_id)
)
''')

print("✅ Все таблицы созданы успешно!")

# @title 💾 Заполнение данными
# @markdown Вставка примеров данных во все таблицы

# Вставка данных в countries
countries_data = [
    ('US', 37.0902, -95.7129, 'USA', 15.5),
    ('FR', 46.2276, 2.2137, 'France', 12.0),
    ('IT', 41.8719, 12.5674, 'Italy', 14.5),
    ('ES', 40.4637, -3.7492, 'Spain', 15.0),
]
cursor.executemany('INSERT INTO countries (country_abbr, latitude, longitude, country_name, temperature) VALUES (?, ?, ?, ?, ?)', countries_data)

# Вставка данных в provinces
provinces_data = [
    ('California', 65.0, 18.3),
    ('Bordeaux', 58.0, 14.4),
    ('Tuscany', 62.0, 16.7),
    ('Rioja', 60.0, 15.6),
    ('Oregon', 55.0, 12.8),
]
cursor.executemany('INSERT INTO provinces (province_name, avg_f, avg_c) VALUES (?, ?, ?)', provinces_data)

# Вставка данных в regions
regions_data = [
    ('Napa Valley',),
    ('Sonoma',),
    ('Willamette Valley',),
    ('Chianti',),
    ('Piedmont',),
]
cursor.executemany('INSERT INTO regions (region_name) VALUES (?)', regions_data)

# Вставка данных в wineries
wineries_data = [
    ('Opus One',),
    ('Chateau Margaux',),
    ('Antinori',),
    ('Vega Sicilia',),
    ('Screaming Eagle',),
    ('Simple Wine Co.',),
]
cursor.executemany('INSERT INTO wineries (winery_name) VALUES (?)', wineries_data)

# Вставка данных в wine_types
wine_types_data = [
    ('Cabernet Sauvignon',),
    ('Merlot',),
    ('Pinot Noir',),
    ('Chardonnay',),
    ('Sangiovese',),
    ('Tempranillo',),
]
cursor.executemany('INSERT INTO wine_types (wine_type) VALUES (?)', wine_types_data)

# Вставка данных в wines
wines_data = [
    ('Opus One 2018', 1, 1, 1, 1, 1, 96, 400.00),
    ('Chateau Margaux 2015', 2, 2, None, 2, 2, 98, 800.00),
    ('Tignanello 2017', 3, 3, 4, 3, 5, 94, 150.00),
    ('Vega Sicilia Unico 2011', 4, 4, None, 4, 6, 97, 500.00),
    ('Screaming Eagle Cabernet 2019', 1, 1, 1, 5, 1, 99, 3500.00),
    ('Simple Wine Co. Table Wine', 1, 5, 3, 6, 3, 75, 12.00),
    ('Budget Bordeaux', 2, 2, None, None, 2, 80, 15.00),
    ('Oregon Pinot Noir Reserve', 1, 5, 3, None, 3, 91, 45.00),
    ('Chianti Classico', 3, 3, 4, None, 5, 88, 25.00),
    ('Napa Valley Chardonnay', 1, 1, 1, None, 4, 90, 55.00),
    ('Barolo Riserva', 3, None, 5, None, 5, 93, 85.00),
    ('Rioja Gran Reserva', 4, 4, None, None, 6, 92, 65.00),
]
cursor.executemany('''INSERT INTO wines (wine_name, country_id, province_id, region_id, winery_id, wine_type_id, rating, price)
                      VALUES (?, ?, ?, ?, ?, ?, ?, ?)''', wines_data)

conn.commit()
print("✅ Данные успешно загружены!")
print(f"   - Стран: {len(countries_data)}")
print(f"   - Регионов: {len(provinces_data)}")
print(f"   - Областей: {len(regions_data)}")
print(f"   - Виноделен: {len(wineries_data)}")
print(f"   - Типов вин: {len(wine_types_data)}")
print(f"   - Вин: {len(wines_data)}")

# @title 📊 Вопрос 1: Самое дорогое и самое дешевое вино
# @markdown Какая страна производила самое дорогое вино? Какое вино было самым дешевым?

query1_expensive = '''
SELECT
    c.country_name as Страна,
    w.wine_name as Вино,
    w.price as Цена
FROM wines w
JOIN countries c ON w.country_id = c.country_id
ORDER BY w.price DESC
LIMIT 1
'''

query1_cheap = '''
SELECT
    c.country_name as Страна,
    w.wine_name as Вино,
    w.price as Цена
FROM wines w
JOIN countries c ON w.country_id = c.country_id
ORDER BY w.price ASC
LIMIT 1
'''

print("💰 САМОЕ ДОРОГОЕ ВИНО:")
display(pd.read_sql_query(query1_expensive, conn))

print("\n💵 САМОЕ ДЕШЕВОЕ ВИНО:")
display(pd.read_sql_query(query1_cheap, conn))

# @title 🌍 Вопрос 2: Самый популярный сорт вина по странам
# @markdown Определите самый популярный сорт вина по странам

query2 = '''
SELECT
    c.country_name as Страна,
    wt.wine_type as Сорт_вина,
    COUNT(*) as Количество
FROM wines w
JOIN countries c ON w.country_id = c.country_id
JOIN wine_types wt ON w.wine_type_id = wt.wine_type_id
GROUP BY c.country_name, wt.wine_type
ORDER BY c.country_name, Количество DESC
'''

print("🍇 ПОПУЛЯРНОСТЬ СОРТОВ ВИНА ПО СТРАНАМ:")
display(pd.read_sql_query(query2, conn))

# @title ⭐ Вопрос 3: Наилучший рейтинг по областям
# @markdown Вина какой области (province) получают наилучший рейтинг? Более дорогие вина оцениваются выше?

query3_rating = '''
SELECT
    p.province_name as Область,
    ROUND(AVG(w.rating), 2) as Средний_рейтинг,
    COUNT(*) as Количество_вин
FROM wines w
JOIN provinces p ON w.province_id = p.province_id
GROUP BY p.province_name
ORDER BY Средний_рейтинг DESC
'''

query3_correlation = '''
SELECT
    ROUND(AVG(rating), 2) as Общий_средний_рейтинг,
    ROUND(AVG(price), 2) as Средняя_цена,
    ROUND(MIN(rating), 2) as Мин_рейтинг,
    ROUND(MAX(rating), 2) as Макс_рейтинг,
    ROUND(MIN(price), 2) as Мин_цена,
    ROUND(MAX(price), 2) as Макс_цена
FROM wines
'''

print("🏆 РЕЙТИНГ ВИНА ПО ОБЛАСТЯМ:")
display(pd.read_sql_query(query3_rating, conn))

print("\n📈 ОБЩАЯ СТАТИСТИКА (корреляция цены и качества):")
display(pd.read_sql_query(query3_correlation, conn))

# @title 🥇 Вопрос 4: Вина с наилучшим рейтингом
# @markdown Определить, какие вины лидируют по рейтингам

query4_top = '''
SELECT
    w.wine_name as Вино,
    w.rating as Рейтинг,
    w.price as Цена,
    c.country_name as Страна,
    p.province_name as Область,
    wt.wine_type as Сорт
FROM wines w
JOIN countries c ON w.country_id = c.country_id
LEFT JOIN provinces p ON w.province_id = p.province_id
JOIN wine_types wt ON w.wine_type_id = wt.wine_type_id
ORDER BY w.rating DESC
LIMIT 5
'''

print("🏅 ТОП-5 ВИН ПО РЕЙТИНГУ:")
display(pd.read_sql_query(query4_top, conn))

# @title 🏭 Вопрос 5: Лучшие винодельни
# @markdown Какие винодельни производят самые дорогие/высоко оцененные вина?

query5_expensive = '''
SELECT
    win.winery_name as Винодельня,
    ROUND(AVG(w.price), 2) as Средняя_цена,
    MAX(w.price) as Макс_цена,
    COUNT(*) as Количество_вин
FROM wines w
JOIN wineries win ON w.winery_id = win.winery_id
GROUP BY win.winery_name
ORDER by Средняя_цена DESC
'''

query5_rating = '''
SELECT
    win.winery_name as Винодельня,
    ROUND(AVG(w.rating), 2) as Средний_рейтинг,
    MAX(w.rating) as Макс_рейтинг,
    COUNT(*) as Количество_вин
FROM wines w
JOIN wineries win ON w.winery_id = win.winery_id
GROUP BY win.winery_name
ORDER BY Средний_рейтинг DESC
'''

print("💎 ВИНОДЕЛЬНИ С САМЫМИ ДОРОГИМИ ВИНАМИ:")
display(pd.read_sql_query(query5_expensive, conn))

print("\n⭐ ВИНОДЕЛЬНИ С НАИВЫСШИМИ РЕЙТИНГАМИ:")
display(pd.read_sql_query(query5_rating, conn))

# @title 📉 Вопрос 6: Вина с низким рейтингом
# @markdown Какие винодельни/регионы/страны производят вина с самым низким рейтингом?

query6_wineries = '''
SELECT
    win.winery_name as Винодельня,
    ROUND(AVG(w.rating), 2) as Средний_рейтинг,
    MIN(w.rating) as Мин_рейтинг,
    COUNT(*) as Количество
FROM wines w
JOIN wineries win ON w.winery_id = win.winery_id
GROUP BY win.winery_name
ORDER BY Средний_рейтинг ASC
LIMIT 5
'''

query6_countries = '''
SELECT
    c.country_name as Страна,
    ROUND(AVG(w.rating), 2) as Средний_рейтинг,
    MIN(w.rating) as Мин_рейтинг,
    MAX(w.rating) as Макс_рейтинг,
    COUNT(*) as Количество
FROM wines w
JOIN countries c ON w.country_id = c.country_id
GROUP BY c.country_name
ORDER BY Средний_рейтинг ASC
'''

print("📉 ВИНОДЕЛЬНИ С НАИЗКИМ РЕЙТИНГОМ:")
display(pd.read_sql_query(query6_wineries, conn))

print("\n🌍 СТРАНЫ ПО СРЕДНЕМУ РЕЙТИНГУ:")
display(pd.read_sql_query(query6_countries, conn))

# @title 📊 Дополнительная статистика
# @markdown Общее соотношение цена/качество

query_extra = '''
SELECT
    w.wine_name as Вино,
    w.rating as Рейтинг,
    w.price as Цена,
    ROUND((w.rating / CASE WHEN w.price > 0 THEN w.price ELSE 1 END) * 100, 2) as Соотношение_цена_качество,
    c.country_name as Страна
FROM wines w
JOIN countries c ON w.country_id = c.country_id
WHERE w.price > 0
ORDER BY Соотношение_цена_качество DESC
LIMIT 10
'''

print("💡 ЛУЧШЕЕ СООТНОШЕНИЕ ЦЕНА/КАЧЕСТВО:")
display(pd.read_sql_query(query_extra, conn))

# Общая статистика
query_stats = '''
SELECT
    COUNT(*) as Всего_вин,
    ROUND(AVG(rating), 2) as Средний_рейтинг,
    MIN(rating) as Мин_рейтинг,
    MAX(rating) as Макс_рейтинг,
    ROUND(AVG(price), 2) as Средняя_цена,
    MIN(price) as Мин_цена,
    MAX(price) as Макс_цена
FROM wines
'''

print("\n📊 ОБЩАЯ СТАТИСТИКА ПО БАЗЕ:")
display(pd.read_sql_query(query_stats, conn))

print("\n✅ Все запросы выполнены успешно!")
```
